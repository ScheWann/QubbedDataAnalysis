```python
import pandas as pd
```

```python
dose_df = pd.read_csv("../data/merged_dose_20260421.csv")
dose_df.drop(columns=['Unnamed: 0'],inplace=True)
dose_df.head()
```

```python
tssim_df = pd.read_csv("../data/tssim_clusters_20260421_original.csv")
tssim_df.rename(columns={"patient_id": "id"}, inplace=True)
tssim_df.head()
```

```python
# Build Anonymized_ID -> STIEFEL number mapping from all three sources
map_files = [
    '../data/Anonymized_DVH_Values____to_be_shared/PatientIDs_merged_Anonymized_INCLUDING_STIEFEL_IDs.xlsx',
    '../data/Anonymized_DVH_Values____to_be_shared-for__Batch_10/PatientIDs_Anonymized_INCLUDING_STIEFEL_IDs.xlsx',
]
id_map_df = pd.concat([pd.read_excel(f) for f in map_files], ignore_index=True)

# merged_sent file uses different column names ('Anonymized', 'ID')
extra_df = pd.read_excel('../data/merged_sent_including_Stiefel_IDs_Anonymized.xlsx')
extra_df = extra_df.rename(columns={'Anonymized': 'Anonymized_ID', 'ID': 'STIEFEL_ID'})
id_map_df = pd.concat([id_map_df, extra_df[['Anonymized_ID', 'STIEFEL_ID']]], ignore_index=True)
id_map_df = id_map_df.drop_duplicates('Anonymized_ID')

id_map_df['stiefel_num'] = id_map_df['STIEFEL_ID'].str.replace('STIEFEL_', '', regex=False).astype(int)
anon_to_stiefel = dict(zip(id_map_df['Anonymized_ID'], id_map_df['stiefel_num']))
print(f'Mapping entries: {len(anon_to_stiefel)}')

# Replace TSSIM df 'id' (Anonymized_ID) with the STIEFEL numeric id (matches dose_df.id)
tssim_df = tssim_df.copy()
tssim_df['id'] = tssim_df['id'].map(anon_to_stiefel)
unmapped = tssim_df['id'].isna().sum()
print(f'TSSIM rows without a STIEFEL mapping: {unmapped}')
tssim_df = tssim_df.dropna(subset=['id'])
tssim_df['id'] = tssim_df['id'].astype(int)

# Compare overlap with dose_df
tssim_ids = set(tssim_df['id'])
dose_ids = set(dose_df['id'])
common = tssim_ids & dose_ids
print(f'Unique ids in TSSIM df: {len(tssim_ids)}')
print(f'Unique ids in dose_df:  {len(dose_ids)}')
print(f'Common ids:             {len(common)}')
print(f'Only in TSSIM df:       {len(tssim_ids - dose_ids)}')
print(f'Only in dose_df:        {len(dose_ids - tssim_ids)}')
```
